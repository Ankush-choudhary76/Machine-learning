```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
from sklearn import datasets
```

```python
iris =datasets.load_iris()
```

```python
iris_data = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
iris_data
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
x_scaler=scaler.fit_transform(iris_data)
x_scaler
```

```python
x_scaler.shape
```

```python
##  4  feature convert into 2 features 
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
pca_scaled = pca.fit_transform(x_scaler)
```

```python
pca_scaled
```

```python
plt.scatter(pca_scaled[:,0],pca_scaled[:,1])
```

```python
## Agglomerative Clustering 
## TO construct a dendogram 

import scipy.cluster.hierarchy as sc

plt.figure(figsize=(20, 7))
plt.title("Dendograms")

## create Dendograms 
sc.dendrogram(sc.linkage(pca_scaled,method='ward'))
plt.title('Dendograms')
plt.xlabel('sample index ')
plt.ylabel('Euclidian Distance ')
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(n_clusters=2 , metric='euclidean',linkage='ward')
cluster.fit(pca_scaled)
```

```python
cluster.labels_
```

```python

plt.scatter(pca_scaled[:,0],pca_scaled[:,1],c=cluster.labels_)
```

```python
from sklearn.metrics import silhouette_score
```

```python
silhouette_coffi = []

for k in range(2,11):
    agglo =AgglomerativeClustering(n_clusters=k , metric='euclidean',linkage='ward')
    agglo.fit(x_scaler)
    score = silhouette_score(x_scaler,agglo.labels_)
    silhouette_coffi.append(score)
```

```python
plt.plot(range(2,11),silhouette_coffi)
plt.xticks(range(2,11))
plt.xlabel("Number of Clusters")
plt.ylabel("silhouette_coefficient")
plt.show()
```
