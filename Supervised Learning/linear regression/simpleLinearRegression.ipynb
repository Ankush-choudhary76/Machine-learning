## Simple Linear Regression 

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

## importing the dataset 

```python
dataset = pd.read_csv("/Users/ankushchoudhari76/Downloads/Salary_Data.csv")
dataset.head()
```

```python
x = dataset.iloc[:,: -1].values
y = dataset.iloc[:,-1].values
```

# Splitting the dataset into the training set and test set 

```python
from sklearn.model_selection import train_test_split
x_train , x_test ,y_train , y_test = train_test_split(x,y,test_size=1/3 , random_state=0)
```

## Training the Sample Linear Regression  model on the training set 

```python
from sklearn.linear_model import LinearRegression
reqressor =  LinearRegression()
reqressor.fit(x_train,y_train)
```

## prediciting the test set result 

```python
y_pred = reqressor.predict(x_test)
```

## Visulising the  Training set result 

```python
plt.scatter(x_train,y_train,color = 'red')
plt.plot(x_train,reqressor.predict(x_train),color = 'blue')
plt.title('salary vs Experience (test set )')
plt.xlabel('year of Experience ')
plt.ylabel('salary')
plt.show
```
